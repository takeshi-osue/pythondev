```python
score_sum = 0
count = 0

print("テストの点数を入力してください（終了する場合は -1 を入力）。")
while True:
    try:
        input_score = int(input("点数を入力してください："))
        
    except ValueError:
        print("⚠︎ 無効な入力です。整数値を入力してください。\n")
        continue

    if input_score == -1:
        break
    
    score_sum += input_score
    count += 1

if count == 0:
    print("\n入力された点数がありませんでした。平均点は計算できません。")
else:
    average_raw = score_sum / count
    
    average_rounded = round(average_raw, 1)

    print(f"\n--- 結果 ---")
    print(f"{count}人のテストの平均点は{average_rounded:.1f}点です。")
```
